```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.cluster as skc
```

```python
dataset = pd.read_csv('../data/Mall_Customers.csv')
X = dataset.iloc[:, 3:5].values
```

```python
wcss = []
for i in range(1, 10):
    kmeans = skc.KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
```

```python
plt.plot(range(1,10), wcss)
plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS (Within Cluster Sum of Squares)')
plt.show()

# Choosing 5 clusters based on elbow method
```

```python
kmeans = skc.KMeans(n_clusters=5, init='k-means++', random_state=42)
y_kmeans = kmeans.fit_predict(X)
```

```python
plt.scatter(X[y_kmeans==0,0], X[y_kmeans==0,1], s=25, color='red', label='Cluster 1')
plt.scatter(X[y_kmeans==1,0], X[y_kmeans==1,1], s=25, color='yellow', label='Cluster 2')
plt.scatter(X[y_kmeans==2,0], X[y_kmeans==2,1], s=25, color='blue', label='Cluster 3')
plt.scatter(X[y_kmeans==3,0], X[y_kmeans==3,1], s=25, color='cyan', label='Cluster 4')
plt.scatter(X[y_kmeans==4,0], X[y_kmeans==4,1], s=25, color='orange', label='Cluster 5')
plt.title('Cloosters')
plt.xlabel('Annual Income (000s)')
plt.ylabel('Spending Score')
plt.legend()
plt.show()
```
