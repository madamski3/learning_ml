```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.cluster as skc
```

```python
dataset = pd.read_csv('../data/Mall_Customers.csv')
X = dataset.iloc[:, 3:5].values
```

```python
import scipy.cluster.hierarchy as sch
dendrogram = sch.dendrogram(sch.linkage(X, method='ward'))
plt.title('Dendrogram')
plt.xlabel('Customers')
plt.ylabel('Euclidean Distances')
plt.show()
```

```python
# affinity = methodology for calculating distance; linkage = which datapoint(s) to use when calculating distance between two clusters
hc = skc.AgglomerativeClustering(n_clusters=5, linkage='ward', metric='euclidean')

y_hc = hc.fit_predict(X)
```

```python
plt.scatter(X[y_hc==0,0], X[y_hc==0,1], s=25, color='red', label='Cluster 1')
plt.scatter(X[y_hc==1,0], X[y_hc==1,1], s=25, color='yellow', label='Cluster 2')
plt.scatter(X[y_hc==2,0], X[y_hc==2,1], s=25, color='blue', label='Cluster 3')
plt.scatter(X[y_hc==3,0], X[y_hc==3,1], s=25, color='cyan', label='Cluster 4')
plt.scatter(X[y_hc==4,0], X[y_hc==4,1], s=25, color='orange', label='Cluster 5')
plt.title('Cloosters (Hierarchical Clustering)')
plt.xlabel('Annual Income (000s)')
plt.ylabel('Spending Score')
plt.legend()
plt.show()
```
