```python
import numpy as np
from tensorflow.keras import models, preprocessing, layers, optimizers, losses, metrics
```

```python
image_path = '../data/cnn_images/'

train_datagen = preprocessing.image.ImageDataGenerator(
    rescale = 1.0 / 255,
    shear_range = 0.2,
    zoom_range = 0.2,
    horizontal_flip = True
)
test_datagen = preprocessing.image.ImageDataGenerator(
    rescale = 1.0 / 255
)

training_set = train_datagen.flow_from_directory(
    image_path + 'training_set',
    target_size = (64, 64),
    batch_size = 32,
    class_mode = 'binary'
)
testing_set = train_datagen.flow_from_directory(
    image_path + 'test_set',
    target_size = (64, 64),
    batch_size = 32,
    class_mode = 'binary'
)
```

```python
cnn_layers = [
    
    # 4 Convolution and pooling layers; first convolution layer takes input
    layers.Conv2D( filters=32, kernel_size=3, activation='relu', input_shape=(64, 64, 3) ),
    layers.MaxPool2D( pool_size=2, strides=2 ),
    layers.Conv2D( filters=32, kernel_size=3, activation='relu'),
    layers.MaxPool2D( pool_size=2, strides=2 ),
    layers.Conv2D( filters=32, kernel_size=3, activation='relu'),
    layers.MaxPool2D( pool_size=2, strides=2 ),
    layers.Conv2D( filters=32, kernel_size=3, activation='relu'),
    layers.MaxPool2D( pool_size=2, strides=2 ),
    
    # Flatten to 2D array
    layers.Flatten(),
    
    # Hidden layers
    layers.Dense(units=128, activation='relu'),
    layers.Dense(units=128, activation='relu'),
    
    # Output layer
    layers.Dense(units=1, activation='sigmoid')
]

cnn = models.Sequential(cnn_layers)
cnn.compile(
    optimizer = optimizers.Adam(),
    loss = losses.BinaryCrossentropy(),
    metrics = [ metrics.BinaryAccuracy() ]
)
```

```python
cnn.fit(x=training_set, validation_data=testing_set, epochs=25)
```

```python
test_image = preprocessing.image.load_img(image_path + 'single_prediction/cat3.png', target_size = (64, 64))
test_image = preprocessing.image.img_to_array(test_image)
test_image = test_image / 255.0
test_image = np.expand_dims(test_image, axis = 0)

result = cnn.predict(test_image)
classification = result[0][0]

print('Probability of dog: ', classification)

if classification >= 0.5:
  print('DOG')
elif classification < 0.5:
  print("CAT")
else:
  print("WHOOPS! YOU SHOULDN'T BE HERE")  
```
