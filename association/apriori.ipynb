```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import apyori as apy
```

```python
dataset = pd.read_csv('../data/Market_Basket_Optimisation.csv', header=None)

dataset_shape = dataset.shape
cnt_rows = dataset_shape[0]
cnt_cols = dataset_shape[1]

transactions = []

# loop over dataset to create a list of items for each individual transaction
for i in range(0, cnt_rows):
    transactions.append([str(dataset.values[i,j]) for j in range(0, cnt_cols)])

print('Cnt transactions=' + str(len(transactions)))
```

```python

min_support = 0.003  # how often does an item need to appear to look for associations (~21/cnt_rows here)
min_confidence = 0.2    # how often must the second item appear for transactions with the first item
min_lift = 3    # required increase in likelihood of second item when we know first item is present
min_length = 2  # min number of items contained in a rule (i.e. when A, then B is more likely vs. when A and B, then C is more likely)
max_length = 2  # (same idea as above)

rules = apy.apriori(transactions=transactions, min_support=min_support, min_confidence=min_confidence, min_lift=min_lift, min_length=min_length, max_length=max_length)
```

```python
results = list(rules)

results
```

```python
def inspect(results):
    lhs         = [tuple(result[2][0][0])[0] for result in results]
    rhs         = [tuple(result[2][0][1])[0] for result in results]
    supports    = [result[1] for result in results]
    confidences = [result[2][0][2] for result in results]
    lifts       = [result[2][0][3] for result in results]
    return list(zip(lhs, rhs, supports, confidences, lifts))
resultsinDataFrame = pd.DataFrame(inspect(results), columns = ['Left Hand Side', 'Right Hand Side', 'Support', 'Confidence', 'Lift'])
```
